```python
from numpy import asarray
from numpy import arange
from numpy.random import rand
from matplotlib import pyplot
import numpy as np
```

```python
def objective(x):
  return x**2.0

def derivative(x):
  return x**2.0

def gradient_descent(objective,derivative,bounds,n_iter, step_size):
  solutions, score=list(),list()
  solution=bounds[:,0]+rand(len(bounds))*(bounds[:,1]-bounds[:,0])
  for i in range(n_iter):
    gradient = derivative(solution)
    solution=solution-step_size * gradient
    solution_eval=objective(solution)
    solutions.append(solution_eval)
    print(f'>{i} f({solution.item()})={solution_eval.item():.5f}')
  return[solutions, score]
```

```python
bounds=asarray([[-1.0,1.0]])
n_iter=30
step_size=0.1

solutions, score=gradient_descent(objective,derivative,bounds,n_iter,step_size)
inputs=arange(bounds[0,0],bounds[0,1]+0.1,0.1)
results=objective(inputs)
print(results)
```

```python
pyplot.plot(inputs, results, 'b', label='Objective')
pyplot.plot(np.ravel(solutions), np.ravel(score), 'r.-', label='Solutions')
pyplot.xlabel('x')
pyplot.ylabel('f(x)')
pyplot.legend()
pyplot.show()
```
